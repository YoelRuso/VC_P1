### Tarea 1 hecha con IA

```python
alto_ajedrez, ancho_ajedrez = 800, 800
celda = 100

tablero = np.zeros((alto_ajedrez, ancho_ajedrez, 3), dtype = np.uint8)

filas, columnas = alto_ajedrez // celda, ancho_ajedrez // celda

for fila in range(filas):
    for columna in range(columnas):
        # Alternamos color según la suma de fila+columna (par -> blanco, impar -> negro)
        if (fila + columna) % 2 == 0:
            x1, y1 = columna * celda, fila * celda
            x2, y2 = x1 + celda, y1 + celda
            cv2.rectangle(tablero, (x1, y1), (x2, y2), (255, 255, 255), -1)
            
print(tablero.shape)
#Visualiza con matplotlib (sin especificar el mapa de color gris)
plt.imshow(tablero) 
plt.show()
```

Evento de ratón para mostrar el valor RGB del píxel

```python
# Función que trata eventos de ratón 
def mouse_events(event, x, y, flags, params):
    global px, py, R, G, B
    px = -1
    # Movimiento del puntero     
    if event==cv2.EVENT_MOUSEMOVE:
        #Valores del píxel, desde la webcam no es BGR
        B = frame[y, x, 0]
        G = frame[y, x, 1]
        R = frame[y, x, 2]  
        # Almacena valores del evento
        px = x
        py = y
```

```python
vid = cv2.VideoCapture(0)

# Tipografía para mostrar texto
font = cv2.FONT_HERSHEY_SIMPLEX
  
lanzado = 0 #bandera para primer fotograma
px = -1
while(True):      
    # fotograma a fotograma
    ret, frame = vid.read()

    if ret: 
        # Activa el manejador en el primer fotograma
        if lanzado == 0:
            # Muestra fotograma
            cv2.imshow('Cam', frame)   
            # Define el nombre del manejador del evento
            cv2.setMouseCallback('Cam', mouse_events)
            lanzado = 1
            
        # Muestra valores RGB intentando centrar en el puntero
        if px > -1:
            cv2.putText(frame, '{}'.format(R), (px-25,py-5), font, 0.5, (0, 0, 255), 1)
            cv2.putText(frame, '     {}'.format(G), (px,py-5), font, 0.5, (0, 255, 0), 1)
            cv2.putText(frame, '         {}'.format(B), (px+25,py-5), font, 0.5, (255, 0, 0), 1)
            
        cv2.imshow('Cam', frame)   
        
    # Detenemos pulsado ESC
    if cv2.waitKey(20) == 27:
        break
  
# Libera el objeto de captura
vid.release()
# Destruye ventanas
cv2.destroyAllWindows()
```

```python
vid = cv2.VideoCapture(0)

#Dimensiones de la cámara
w = int(vid.get(cv2.CAP_PROP_FRAME_WIDTH))
h = int(vid.get(cv2.CAP_PROP_FRAME_HEIGHT))

#Fuerzo a mitad de resolución para ocupar menos pantalla
w=int(w/2)
h=int(h/2)
vid.set(cv2.CAP_PROP_FRAME_WIDTH, w) #En Mac no reacciona a estos comandos
vid.set(cv2.CAP_PROP_FRAME_HEIGHT, h)

#Imagen conjunta 2x original
collage = np.zeros((h*2,w*2,3), dtype = np.uint8)
tl = collage[0:h,0:w]
tr = collage[0:h,w:w+w]
bl = collage[h:h+h,0:w]
br = collage[h:h+h,w:w+w]

while True:      
    # fotograma a fotograma
    ret, frameIN = vid.read()

    #Menor tamaño
    frame = cv2.resize(frameIN, (int(w),int(h)),cv2.INTER_NEAREST)

    if ret:
        #Separamos canales
        r = frame[:,:,2]
        g = frame[:,:,1]
        b = frame[:,:,0]

        #Jugamos con los valores de los planos
        tl[:,:,0] = b
        tl[:,:,1] = g
        tl[:,:,2] = r

        tr[:,:,0] = 255 - r
        tr[:,:,1] = 255 - g
        tr[:,:,2] = b
        
        bl[:,:,0] = r
        bl[:,:,1] = 255 - b
        bl[:,:,2] = g

        br[:,:,0] = b
        br[:,:,1] = g
        br[:,:,2] = 255 - r
    
        # Muestra composición
        cv2.imshow('Cam', collage)
    
    # Detenemos pulsado ESC
    if cv2.waitKey(20) == 27:
        break
  
# Libera el objeto de captura
vid.release()
# Destruye ventanas
cv2.destroyAllWindows()
```

Transformando la imagen capturada en una colección de círculos, cuyo radio depende de la intensidad promedio de la zona

```python
vid = cv2.VideoCapture(0)

# Se define el número de celdas
ncells = 10
off = int(ncells/2)
  
while(True):      
    # fotograma a fotograma
    ret, frame = vid.read()

    if ret:
        #Dimensiones originales
        h, w, c = frame.shape
        #Redimensiono teniendo en cuenta ncells
        down_frame = cv2.resize(frame, (int(w/ncells),int(h/ncells)),cv2.INTER_NEAREST)
        #Dimensiones reducidas
        h2, w2, c2 = down_frame.shape

        #Separamos canales
        r = down_frame[:,:,0]
        g = down_frame[:,:,1]
        b = down_frame[:,:,2]

        # Creamos imagen negra
        gris_up_frame = np.zeros((h2*ncells,w2*ncells,1), dtype = np.uint8)

        # recorre valores de la imagen reducida
        for y in range(0,h2):
            for x in range(0,w2):
                # La suma de los valores RGB define el radio del círculo
                rad = int ( (r[y,x] + g[y,x] + b[y,x])/(ncells*3*2))
                cv2.circle(gris_up_frame,(x*ncells+off, y*ncells + off), rad, (255,255,255), -1)

    
        # Muestra fotograma resultante
        cv2.imshow('Cam', gris_up_frame)
    
    # Detenemos pulsado ESC
    if cv2.waitKey(20) == 27:
        break
  
# Libera el objeto de captura
vid.release()
# Destruye ventanas
cv2.destroyAllWindows()
```

```python
"""
Pop Art estilo Warhol con la webcam.
Collage 2x2: la misma imagen posterizada a 4 tonos con 4 paletas distintas.
Uno de los niveles muestra un patron de fondo, recoloreado en cada cuadrante.

Teclas:
  ESC   salir
  f     activa o desactiva el patron de fondo
  0..3  elige que nivel de tono muestra el patron (0 = mas oscuro, 3 = mas claro)
  p     rota que combinacion de colores va a cada cuadrante
  m     alterna el modo de recoloreado: dos tonos / tinte
  a     alterna umbrales automaticos (percentiles) / fijos
  c     activa o desactiva la linea de contorno
  g     activa o desactiva el realce de contraste (CLAHE)
  + -   desplaza los umbrales (mas claro / mas oscuro)
  s     guarda un PNG del collage
"""

import cv2
import numpy as np

# ----------------------- AJUSTES -----------------------
IMAGEN_FONDO = 'pop-art-gary-grayson.jpg'
NIVEL_IMAGEN = 0                 # 0 = tono mas oscuro, 3 = mas claro
VARIAR_FONDO = True              # voltea el patron en cada cuadrante

# 'dos_tonos': separa el patron en fondo y motivo y los repinta (ideal para
#              tramas planas: estrellas, lunares, rayas)
# 'tinte':     desplaza el tono en HSV, sirve para cualquier imagen
MODO_PATRON = 'dos_tonos'

# Un par (color_fondo, color_motivo) por cuadrante, en BGR.
COLORES_PATRON = [
    ((  0, 240, 255), (150,  20, 230)),   # amarillo + estrellas magenta
    ((255, 225, 140), (180,  60,   0)),   # azul claro + estrellas azul fuerte
    ((140, 235, 255), ( 40,  40, 205)),   # crema + estrellas rojas
    ((190, 255, 190), ( 60, 160,  40)),   # verde claro + estrellas verde oscuro
]
INVERTIR_PATRON = False          # True si el motivo sale intercambiado con el fondo

DESPLAZAMIENTO_TONO = [0, 45, 90, 135]   # solo para MODO_PATRON = 'tinte'
MEZCLA_COLOR = 0.20                      # solo para MODO_PATRON = 'tinte'

AUTO_NIVELES   = True
PERCENTILES    = (22, 50, 78)
UMBRALES_FIJOS = (60, 110, 170)
CONTORNOS      = True
CLAHE          = True

# Paletas en BGR, ordenadas de tono OSCURO a tono CLARO.
PALETAS = [
    [(35,  15,  45), (150,  20, 230), (195, 155, 255), (0, 235, 255)],  # rosa
    [(30,  25,  15), (165, 170,  25), (190, 160, 240), (0, 240, 255)],  # turquesa
    [(25,  20,  15), ( 20, 140, 255), (200, 180, 250), (0, 240, 255)],  # naranja
    [(20,  15,  15), ( 40,  40, 205), (180, 190, 250), (0, 235, 255)],  # rojo
]


def crear_lut(paleta, umbrales):
    t1, t2, t3 = umbrales
    lut = np.zeros((256, 1, 3), dtype=np.uint8)
    lut[0:t1]   = paleta[0]
    lut[t1:t2]  = paleta[1]
    lut[t2:t3]  = paleta[2]
    lut[t3:256] = paleta[3]
    return lut


def calcular_umbrales(gris, auto, desplazamiento):
    if auto:
        t1, t2, t3 = np.percentile(gris, PERCENTILES)
    else:
        t1, t2, t3 = UMBRALES_FIJOS
    t = [int(np.clip(int(v) + desplazamiento, 1, 255)) for v in (t1, t2, t3)]
    t[1] = max(t[1], t[0] + 1)
    t[2] = max(t[2], t[1] + 1)
    return [min(v, 255) for v in t]


def cargar_imagen(ruta):
    """Lee la imagen. Si OpenCV no trae soporte webp, prueba con Pillow."""
    img = cv2.imread(ruta, cv2.IMREAD_COLOR)
    if img is not None:
        return img
    try:
        from PIL import Image
        pil = Image.open(ruta).convert('RGB')
        return cv2.cvtColor(np.array(pil), cv2.COLOR_RGB2BGR)
    except Exception:
        return None


def encajar(img, w, h):
    """Recorte centrado a la proporcion del cuadrante y reescalado, sin deformar."""
    ih, iw = img.shape[:2]
    escala = max(w / iw, h / ih)
    nw, nh = int(np.ceil(iw * escala)), int(np.ceil(ih * escala))
    img = cv2.resize(img, (nw, nh), interpolation=cv2.INTER_AREA)
    x0, y0 = (nw - w) // 2, (nh - h) // 2
    return img[y0:y0 + h, x0:x0 + w]


def separar_motivo(img):
    """Devuelve la mascara del motivo (la parte oscura del patron) con Otsu."""
    gris = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    _, binaria = cv2.threshold(gris, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)
    mascara = binaria > 0
    return ~mascara if INVERTIR_PATRON else mascara


def tintar(img, grados, color_paleta):
    """Desplaza el tono en HSV y mezcla un poco con el color del cuadrante."""
    hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
    hsv[:, :, 0] = (hsv[:, :, 0].astype(np.int16) + grados // 2) % 180
    salida = cv2.cvtColor(hsv, cv2.COLOR_HSV2BGR)
    capa = np.full(img.shape, color_paleta, dtype=np.uint8)
    return cv2.addWeighted(salida, 1 - MEZCLA_COLOR, capa, MEZCLA_COLOR, 0)


# ----------------------- CAMARA -----------------------
vid = cv2.VideoCapture(0)

w = int(vid.get(cv2.CAP_PROP_FRAME_WIDTH))
h = int(vid.get(cv2.CAP_PROP_FRAME_HEIGHT))
w = int(w / 2)
h = int(h / 2)
vid.set(cv2.CAP_PROP_FRAME_WIDTH, w)
vid.set(cv2.CAP_PROP_FRAME_HEIGHT, h)

collage = np.zeros((h * 2, w * 2, 3), dtype=np.uint8)
cuadrantes = [
    collage[0:h,     0:w],      # arriba izquierda
    collage[0:h,     w:w + w],  # arriba derecha
    collage[h:h + h, 0:w],      # abajo izquierda
    collage[h:h + h, w:w + w],  # abajo derecha
]

# ----------------------- PATRON DE FONDO -----------------------
fondo_src = cargar_imagen(IMAGEN_FONDO)
hay_patron = fondo_src is not None
if not hay_patron:
    print(f'No se pudo abrir "{IMAGEN_FONDO}". Se usara color plano.')
    volteos, mascaras = [], []
else:
    base = encajar(fondo_src, w, h)
    mascara_base = separar_motivo(base)
    if VARIAR_FONDO:
        codigos = [None, 1, 0, -1]   # sin voltear, horizontal, vertical, ambos
    else:
        codigos = [None] * 4
    volteos, mascaras = [], []
    for cod in codigos:
        volteos.append(base if cod is None else cv2.flip(base, cod))
        m = mascara_base.astype(np.uint8)
        mascaras.append((m if cod is None else cv2.flip(m, cod)) > 0)


def preparar_fondos(nivel, rotacion, modo):
    """Genera el patron ya recoloreado para cada cuadrante. Se llama solo al
    cambiar de ajuste, no en cada fotograma."""
    if not hay_patron:
        return [None] * 4
    salida = []
    for i in range(4):
        j = (i + rotacion) % 4
        if modo == 'dos_tonos':
            color_fondo, color_motivo = COLORES_PATRON[j]
            img = np.full((h, w, 3), color_fondo, dtype=np.uint8)
            img[mascaras[i]] = color_motivo
        else:
            img = tintar(volteos[i], DESPLAZAMIENTO_TONO[j], PALETAS[i][nivel])
        salida.append(img)
    return salida


clahe = cv2.createCLAHE(clipLimit=2.5, tileGridSize=(8, 8))
desplazamiento = 0
auto = AUTO_NIVELES
contornos = CONTORNOS
usar_clahe = CLAHE
usar_fondo = hay_patron
nivel_img = NIVEL_IMAGEN
rotacion = 0
modo = MODO_PATRON
fondos = preparar_fondos(nivel_img, rotacion, modo)
contador = 0

while True:
    ret, frameIN = vid.read()
    if not ret:
        break

    frame = cv2.resize(frameIN, (w, h), interpolation=cv2.INTER_AREA)
    frame = cv2.flip(frame, 1)   # efecto espejo

    gris = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    if usar_clahe:
        gris = clahe.apply(gris)

    gris_plano = cv2.bilateralFilter(gris, 7, 75, 75)
    gris_plano = cv2.medianBlur(gris_plano, 5)

    if contornos:
        bordes = cv2.Canny(cv2.GaussianBlur(gris, (5, 5), 0), 60, 130)
        bordes = cv2.dilate(bordes, np.ones((2, 2), np.uint8))
        mascara_bordes = bordes > 0
    else:
        mascara_bordes = None

    umbrales = calcular_umbrales(gris_plano, auto, desplazamiento)
    gris3 = cv2.cvtColor(gris_plano, cv2.COLOR_GRAY2BGR)

    # mapa de niveles: 0,1,2,3 segun en que franja de tono cae cada pixel
    if usar_fondo:
        mascara_img = np.digitize(gris_plano, umbrales) == nivel_img
    else:
        mascara_img = None

    for i, (cuadrante, paleta) in enumerate(zip(cuadrantes, PALETAS)):
        coloreado = cv2.LUT(gris3, crear_lut(paleta, umbrales))
        if mascara_img is not None:
            coloreado[mascara_img] = fondos[i][mascara_img]
        if mascara_bordes is not None:
            coloreado[mascara_bordes] = (0, 0, 0)
        cuadrante[:, :] = coloreado

    cv2.imshow('Pop Art', collage)

    # ----------------------- TECLAS -----------------------
    k = cv2.waitKey(20) & 0xFF
    if k == 27:
        break
    elif k == ord('f') and hay_patron:
        usar_fondo = not usar_fondo
    elif k in (ord('0'), ord('1'), ord('2'), ord('3')):
        nivel_img = k - ord('0')
        fondos = preparar_fondos(nivel_img, rotacion, modo)
    elif k == ord('p'):
        rotacion = (rotacion + 1) % 4
        fondos = preparar_fondos(nivel_img, rotacion, modo)
    elif k == ord('m'):
        modo = 'tinte' if modo == 'dos_tonos' else 'dos_tonos'
        fondos = preparar_fondos(nivel_img, rotacion, modo)
    elif k == ord('a'):
        auto = not auto
    elif k == ord('c'):
        contornos = not contornos
    elif k == ord('g'):
        usar_clahe = not usar_clahe
    elif k in (ord('+'), ord('=')):
        desplazamiento += 5
    elif k in (ord('-'), ord('_')):
        desplazamiento -= 5
    elif k == ord('s'):
        contador += 1
        cv2.imwrite(f'popart_{contador:02d}.png', collage)
        print(f'Guardado popart_{contador:02d}.png')

vid.release()
cv2.destroyAllWindows()
```
